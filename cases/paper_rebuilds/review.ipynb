# FigureFlow · 九张论文图的原图与复刻
在空白 Colab runtime 运行以下两格。第一格从 FigureFlow 官方仓库取得代码并安装渲染依赖；第二格显式下载六份来源锁定的公开论文，在本次 runtime 本地生成九张对照。不需要模型、GPU 或 API Key。没有预存论文图像。若会议站点限制下载，可手动放入指定目录中的同版本 PDF，再重跑第二格。
渲染使用本机字体，Linux 与 macOS 的文字宽度、阴影、抗锯齿可能略有差异。完整代码、图号、配色、还原边界见同目录 README。

```python
from pathlib import Path
import subprocess, sys
root = Path('FigureFlow').resolve()
if not root.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/Sunbeam23333/FigureFlow.git', str(root)], check=True)
case = root / 'cases/paper_rebuilds'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(case / 'requirements.txt')], check=True)
```

```python
from IPython.display import display, Image, FileLink
papers = case / 'local/sources'
out = case / 'local'
subprocess.run([sys.executable, str(case / 'fetch_sources.py'), '--all', '--source-dir', str(papers)], check=True)
subprocess.run([sys.executable, str(case / 'render.py'), '--all', '--source-dir', str(papers), '--output-dir', str(out)], check=True)
for key in ('videoworld_fig1','janus_fig1','janus_fig2','epona_fig1','epona_fig2','graphmimic_fig1','graphmimic_fig2','motus_fig1','dynaview_fig1'):
    display(Image(filename=str(out / key / 'comparison.png')))
display(FileLink(str((out / 'figureflow_comparisons.pdf').relative_to(Path.cwd()))))
```
